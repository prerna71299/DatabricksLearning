```python
data=[(1,'abc@gmail.com'),(2,'xyz@gmail.com'),(3,'abc@gmail.com'),(4,'pqr@gmail.com'),(5,'xyz@gmail.com')]
column=['id','email']
df=spark.createDataFrame(data=data,schema=column)
df.show()
```

```python
from pyspark.sql.functions import col
display(df.groupBy("email").count().filter(col('count')>1))
```

```python
df.createOrReplaceTempView("df_email")


```

```python
%sql
select * from df_email
```

```python
%sql
select count(*),email from df_email group By email having count(*)>1
```

```python
%sql
select distinct (email) from df_email
```
