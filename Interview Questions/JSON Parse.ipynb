```python
#1. WE are getting data in the form of string JSON we need to convert into json format using pyspark
 #2. After converting we need to create seperate column from the json body.
#using pyspark 
```

```python
data=[("Manish",'{"Street":"123 Main Street","City":"New York"}'),("Ravi",'{"Street":"456 Oak Avenue","City":"Los Angeles"}')]
schema=["name","address"]
df=spark.createDataFrame(data,schema)
df.printSchema()
display(df)

```

```python
df.createOrReplaceTempView("df_view")
#1. WE are getting data in the form of string JSON we need to convert into json format using pyspark
 #2. After converting we need to create seperate column from the json body.
#using SQL
```

```python
%sql
select * from df_view
```

```python
%sql
with cte as (
  select name, address, from_json(address, 'Street string,City string') as address_new from df_view
)
select name, address,address_new.Street as street,address_new.City as city from cte
```

```python
from pyspark.sql.functions import *

df1=df.withColumn("address_new",from_json(col('address'),'Street string,City string'))
display(df1)
```

```python
df1.select('name','address','address_new',col('address_new').Street.alias('Street'),col('address_new').City.alias('City')).display()
```
