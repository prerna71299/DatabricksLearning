```python
#Cumulative (running total) sales
#Previous day sales
#Next day sales
#Using PySpark AND Spark SQL
```

```python
data=[['2024-01-01',20000],['2024-01-02',10000],['2024-01-03',30000],['2024-01-04',40000],['2024-01-05',50000]]
column=['date','sales']
df=spark.createDataFrame(data,column)
display(df)
```

```python
df.createOrReplaceTempView("sample_sales")
```

```python
%sql
select * from sample_sales
```

```python
%sql
select *,sum(sales)over(order by date asc) cumm_sales,lag(sales)over(order by date asc) prev_sales ,lead(sales) over(order by date asc)Next_sales from sample_sales
```

```python
from pyspark.sql.functions import *
from pyspark.sql import Window

window=Window.orderBy('date')
df1=df.withColumn('cumm_sales',sum(col('sales')).over(window))
df1.withColumn('Prev_sales',lag(col('sales')).over(window)).show()
df1.withColumn('Next_sales',lead(col('sales')).over(window)).show()
```
