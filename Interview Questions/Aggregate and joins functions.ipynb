```python
"""PYSPARK INTERVIEW QUESTION 
WE HAVE 2 TABLE EMPLOYEE AND DEPARTMENT TABLE IS GIVEN 

1. WE HAVE TO FIND THE HIGHEST SALARY BASED ON EACH DEPARTMENT NAME
2. WE HAVE TO FIND THE EMPLOYEE WHO IS GETTING HIGHEST SALARY BASED ON EACH DEPARTMENT NAME
3. WE HAVE TO FIND THE lowest SALARY BASED ON EACH DEPARTMENT NAME
4. WE HAVE TO FIND THE EMPLOYEE WHO IS GETTING lowest SALARY BASED ON EACH DEPARTMENT NAME """
```

```python
emp_data=[('Prerana',1,650000),
          ('Ashish',2,9000000),
          ('Abc',1,39000),
          ('Xyz',2,4000000),
          ('Pqr',3,5000000),
          ('Mno',3,6000000),
          ('Prajkta',3,7000000),
          ('Manihsuwi',3,8000000),
          ('Pratil',3,9000000),
          ('Mamnis',3,10000)
          ]
Schema1=["Emp_name","Dept_id","salary"]

emp_df=spark.createDataFrame(emp_data,Schema1)
emp_df.show()

Dept_data=[(1,'IT'),
          (2,'HR'),
          (3,'FINANCE')]
Schema2=["Dept_id","Dept_name"]

Dept_df=spark.createDataFrame(Dept_data,Schema2)
Dept_df.show()


```

```python
from pyspark.sql.functions import *

df1=emp_df.join(Dept_df,emp_df.Dept_id==Dept_df.Dept_id,"left")
df1.show()
```

```python
from pyspark.sql.functions import *
df2=df1.groupBy("Dept_name").agg(max('salary').alias('Max_Salary'))
df2.show()
```

```python
df3=df1.join(df2,df1.Dept_name==df2.Dept_name,"left").filter(df1.salary==df2.Max_Salary)
df3.show()
```

```python
from pyspark.sql.functions import *
df2_min=df1.groupBy("Dept_name").agg(min('salary').alias('Min_Salary'))
df2_min.show()
```

```python
df3_min=df1.join(df2_min,df1.Dept_name==df2_min.Dept_name,"left").filter(df1.salary==df2_min.Min_Salary)
df3_min.show()
```
