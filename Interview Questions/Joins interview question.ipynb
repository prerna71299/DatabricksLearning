```python
# INTERVIE QUESTION 
# WE HAVE 2 TABLE CUSTOMER AND ORDER TABLE IS GIVEN 

# 1. WE HAVE TO FIND OUT CUSTOMER WHO HAVE NOT ORDER ANYTHING 
# 2. WE HAVE TO FIND OUT CUSTOMER WHO HAVE ORDERED 
```

```python
customer_data=[(1,"Manish"),(2,"Ravi"),(3,"Rajesh"),(4,"Rakesh")]
schema_cust=["Customer_ID","Customer_Name"]

df_cust=spark.createDataFrame(data=customer_data,schema=schema_cust)
df_cust.show()

oder_data=[(1,4),(3,2)]
Order_schema=["Order_ID","Customer_ID"]

df_order=spark.createDataFrame(data=oder_data,schema=Order_schema)
df_order.show()

```

```python
display(df_cust.join(df_order,df_cust.Customer_ID==df_order.Customer_ID,"left").filter(df_order.Order_ID.isNull()))
```

```python
display(df_cust.join(df_order,df_cust.Customer_ID==df_order.Customer_ID,"inner"))
```

```python
df_cust.createOrReplaceTempView("customer")
df_order.createOrReplaceTempView("order")

```
