```sql
%python
from datetime import datetime, timedelta

dbutils.widgets.dropdown("time_period","weekly",["Daily","weekly","monthly"])
time_period=dbutils.widgets.get("time_period")

today_datetime=datetime.strptime('2023-11-11', '%Y-%m-%d')
today=today_datetime.date()

if time_period=='weekly':
    start_date=today-timedelta(days=today.weekday(),weeks=1)-timedelta(days=1)
    end_date=start_date+timedelta(days=6)
elif time_period=='monthly':
    first=today.replace(day=1)
    end_date=first-timedelta(days=1)
    start_date=first-timedelta(days=end_date.day)
else:
    start_date=today-timedelta(days=1)
    end_date=start_date


display(spark.sql(f"""select sum(sale),city from retail_data.default.sales  where date between '{start_date}' and '{end_date}' group by city"""))
print(start_date,end_date)
```
