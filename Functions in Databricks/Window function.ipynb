```python
data1=[("Prerana","IT","27","60000"),
       ("Ashish","IT","30","800000"),
       ("pqr","Sales","20","32322"),
       ("xyz","Sales","25","343434"),
       ("abc","Sales","28","30000"),
       ("def","Sales","29","30000"),
    ]

columns=["Name","Dept","Age","Salary"]

df=spark.createDataFrame(data=data1,schema=columns)
df.show()

```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number,dense_rank,rank

window=Window.partitionBy("Dept").orderBy("Salary")
df.withColumn("rn", row_number().over(window)).show()
df.withColumn("rank", rank().over(window)).show()
df.withColumn("dense rank", dense_rank().over(window)).show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number,dense_rank,rank,col

window=Window.partitionBy("Dept").orderBy(col("Salary").desc())
df.withColumn("rn", row_number().over(window)).show()
df.withColumn("rank", rank().over(window)).show()
df.withColumn("dense rank", dense_rank().over(window)).show()
```
