```python
data=[("2022-03-15","2022-03-16 12:10:24"),
      ("2023-03-15","2022-03-16 12:10:24")
      ]
df=spark.createDataFrame(data,["date_col","timest"])
df.show()     
```

```python
from pyspark.sql.functions import *
df.select("date_col",date_add("date_col",30)).show()
```

```python
df1=df.select("date_col",date_sub("date_col",3).alias("date_sub1"))
```

```python
df1.show()
```

```python
df1.select("date_col","date_sub1",datediff("date_sub1","date_col")).show()
```

```python
df1.select("date_col",year("date_col")).show()
```

```python
df1.select("date_col",month("date_col")).show()
```

```python
df1.select("date_col",dayofyear("date_col")).show()
```
