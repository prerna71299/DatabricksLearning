```python
import pyspark
from pyspark.sql import SparkSession

data1=[("Prerana","IT","27","60000"),\
       ("Ashish","IT","30","800000"),\
       ("pqr","Sales","20","32322")
    ]

columns=["Name","Dept","Age","Salary"]

df=spark.createDataFrame(data=data1,schema=columns)
df.show()
```

```python
data2=[("xyz","IT","29","6000"),\
       ("Ash","Sales","40","8000"),\
       ("ewq","Finance","60","2322"),\
        ("pqr","Sales","20","32322")
    ]
columns2=["Name","Dept","Age","Salary"]
df2=spark.createDataFrame(data=data2,schema=columns2)
df2.show()
```

```python
df.union(df2).show()
```

```python
df.unionAll(df2).show()
```

```python
df.union(df2).distinct().show()
```
