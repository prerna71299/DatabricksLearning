```python
data=[(1,"abc",1200),
      (2,"pqr",2000),
      (3,"xyz",5000)
      ]
columns=["id","name","salary"] 
df=spark.createDataFrame(data,columns)
df.show()
     
```

```python
df.createOrReplaceTempView("Employee")
```

```python
%sql
select * from Employee
```

```python
%sql
select max(salary) from Employee
```

```python
%sql
select upper(name),salary from Employee
```
