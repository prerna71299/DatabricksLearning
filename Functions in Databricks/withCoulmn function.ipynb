```python
df=spark.read.csv("/Volumes/samplepractical/schema1/volume1/s1/employees.csv",header=True)
```

```python
df.show()
```

```python
df.withColumn("Incremented Salary",df.Salary*10).show()
```

```python
from pyspark.sql.functions import lit
df.withColumn("Country",lit("UK")).show()
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col
df1=df.withColumn("Salary",col("Salary").cast("Integer"))
```

```python
df1.show()
```

```python
df1.printSchema()
```

```python
df1.withColumnRenamed("Salary","New Salary").show()
```
