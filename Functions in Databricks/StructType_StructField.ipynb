```python
data=[(1,"abc",1200),(2,"def",1300),(3,"ghi",1400)]

columns=["id","name","salary"]

df=spark.createDataFrame(data,columns)
df.show()
```

```python
df.printSchema() #pyspark will generate schema for you based on data but for real world data, you will have to manually specify schema to prevent data type mismatch
```

```python
from pyspark.sql.types import StructField,StringType,IntegerType,StructType

data=[(1,"abc",1200),(2,"def",1300),(3,"ghi",1400)]

schema=StructType([StructField(name='ID',dataType=IntegerType()),
                   StructField(name='Name',dataType=StringType()),
                   StructField(name='Salary',dataType=IntegerType())
                   ])
df=spark.createDataFrame(data,schema)
df.show()
df.printSchema()
```
