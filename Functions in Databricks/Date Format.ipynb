```python
data=[("2024-03-15","2024-03-16 12:12:56.234"),
      ("2024-04-15","2024-04-16 11:11:55.234"),
      ("2024-05-15","2024-05-16 10:10:54.234"),
      ]

df=spark.createDataFrame(data,["date_col","timestamp_col"])
df.show()
```

```python
from pyspark.sql.functions import *

df.select("date_col",date_format("date_col","yyyy/mm/dd").alias("Date")).show()
```

```python
df.select("date_col",date_format("date_col","yyyy/MM/DD")).show() #we have not use alias function so in column name it is showing like this
```
