```python
data=[("abc",["python","C","C++"]),
      ("pqr",["C","C++","C#"]),
      ("xyz",["python","C"])
      ]
columns=["name","Languages"] 
df=spark.createDataFrame(data,columns)
df.show()     
```

```python
from pyspark.sql.functions import *
df.select("name",explode(col("Languages"))).show()

```

```python
df.createOrReplaceTempView("sample")
```

```python
%sql
select * from sample
```

```python
%sql
select name,explode(Languages) from sample
```
